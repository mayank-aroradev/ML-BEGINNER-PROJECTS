```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC
```

```python
db=pd.read_csv('/content/diabetes.csv')
```

```python
features=db.drop(columns='Outcome',axis=1)
target=db['Outcome']
```

```python
classifier=SVC(kernel='linear')
```

```python
scaler=StandardScaler()
scaler.fit(features)
```

```python
std_data=scaler.transform(features)
```

```python
x_train,x_test,y_train,y_test=train_test_split(std_data,target,test_size=0.2,stratify=target,random_state=2)
```

```python
classifier.fit(x_train,y_train)
```

```python
x_train_prediction=classifier.predict(x_train)
training_data_accuracy=accuracy_score(y_train,x_train_prediction)
```

```python
training_data_accuracy
```

```python
x_test_prediction=classifier.predict(x_test)
test_data_accuracy=accuracy_score(y_test,x_test_prediction)
```

```python
test_data_accuracy
```

```python
input_data=(4,110,92,0,0,37.6,0.191,30)
input_dataasnumpy=np.asarray(input_data)
inputdatareshape=input_dataasnumpy.reshape(1,-1)
std_data=scaler.transform(inputdatareshape)

prediction=classifier.predict(std_data)
print(prediction)
```
